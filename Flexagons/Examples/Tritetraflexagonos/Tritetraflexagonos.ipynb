# Geração de Planificações: Tritetraflexágono (3 Faces)
**Projeto EnIMaR 2026 — Coleção Matemática Fora da Caixa / UFMG**

O **tritetraflexágono** é um flexágono quadrado constituído por 3 faces quadradas completas:
- **Face Frente** (`imgF`): visível na configuração inicial frontal.
- **Face Verso** (`imgV`): visível na configuração inicial posterior.
- **Face Escondida** (`imgE`): interiorizada na dobradura, revelada após flexão de $180^\circ$.

Sua montagem é realizada a partir de uma tira de 6 quadrados conectada em octomino. As imagens são divididas em 4 quadrantes cada e redistribuídas entre o plano frontal e o plano traseiro com as orientações adequadas.

Neste notebook:
1. Carregamos e exibimos as 3 imagens de entrada da pasta local `imgs/`.
2. Geramos a planificação **sem marcas de corte** (`grafica=False`).
3. Geramos a planificação **com marcas de corte e facas para gráfica** (`grafica=True`).
4. Apresentamos uma visão comparativa dos planos resultantes.


```python
import os
import sys
from pathlib import Path
import matplotlib.pyplot as plt
from PIL import Image

# Configuração de caminhos do projeto
NOTEBOOK_DIR = Path.cwd().resolve()
PROJECT_ROOT = NOTEBOOK_DIR.parents[2]
FLEXAGONS_DIR = PROJECT_ROOT / "Flexagons"

for p in [str(PROJECT_ROOT), str(FLEXAGONS_DIR), str(FLEXAGONS_DIR / "Tritetraflexagonos")]:
    if p not in sys.path:
        sys.path.insert(0, p)

from Tritetraflexagonos.tritetraflexagonos import plano

IMG_DIR = NOTEBOOK_DIR / "imgs"
print(f"Diretório de trabalho: {NOTEBOOK_DIR}")
print(f"Subpasta de imagens:  {IMG_DIR}")
print(f"Arquivos em imgs/:    {sorted(os.listdir(IMG_DIR))}")
```

## 1. Visualização das 3 Faces de Entrada (`imgs/`)
Carregamento das faces: Frente, Verso e Escondida.

```python
face_frente_path = IMG_DIR / "face_frontal.png"
face_verso_path = IMG_DIR / "face_traseira.png"
face_escondida_path = IMG_DIR / "face_escondida.png"

faces = {
    "Face Frente (imgF)": face_frente_path,
    "Face Verso (imgV)": face_verso_path,
    "Face Escondida (imgE)": face_escondida_path
}

fig, axes = plt.subplots(1, 3, figsize=(16, 6))

for ax, (label, path) in zip(axes, faces.items()):
    img = Image.open(path)
    ax.imshow(img)
    ax.set_title(f"{label}\n{path.name}\nDimensão: {img.width} x {img.height} px", fontsize=11, fontweight='bold')
    ax.axis('off')

plt.tight_layout()
plt.show()
```

## 2. Geração do Plano Sem Marcas de Corte (`grafica=False`)
Gera a planificação direta ($2922 \times 1239$ pixels) sem sangrias estendidas, adequada para visualização digital e corte simples.

```python
print("Gerando planos sem marcas de corte...")
plano_frontal_sem, plano_traseiro_sem = plano(
    imgF=str(face_frente_path),
    imgV=str(face_verso_path),
    imgE=str(face_escondida_path),
    grafica=False,
    nome_arq="sem_marcas"
)

print(f"-> Plano Frontal (sem marcas):  {plano_frontal_sem.size} pixels")
print(f"-> Plano Traseiro (sem marcas): {plano_traseiro_sem.size} pixels")

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(16, 11))

ax1.imshow(plano_frontal_sem)
ax1.set_title(f"Plano Frontal (Sem Marcas de Corte) — {plano_frontal_sem.width} x {plano_frontal_sem.height} px", fontsize=13, fontweight='bold')
ax1.axis('on')
ax1.tick_params(labelsize=8)

ax2.imshow(plano_traseiro_sem)
ax2.set_title(f"Plano Traseiro (Sem Marcas de Corte) — {plano_traseiro_sem.width} x {plano_traseiro_sem.height} px", fontsize=13, fontweight='bold')
ax2.axis('on')
ax2.tick_params(labelsize=8)

plt.suptitle("Tritetraflexágono — Planificação Sem Marcas de Corte", fontsize=15, fontweight='bold')
plt.tight_layout()
plt.show()
```

## 3. Geração do Plano Com Marcas de Corte para Gráfica (`grafica=True`)
Gera a planificação gráfica com sangrias de 58 px e aplicação das facas/marcas de corte de `Cortes/` ($2980 \times 1297$ pixels).

```python
print("Gerando planos com marcas de corte para gráfica...")
plano_frontal_com, plano_traseiro_com = plano(
    imgF=str(face_frente_path),
    imgV=str(face_verso_path),
    imgE=str(face_escondida_path),
    grafica=True,
    nome_arq="com_marcas"
)

print(f"-> Plano Frontal (com marcas):  {plano_frontal_com.size} pixels")
print(f"-> Plano Traseiro (com marcas): {plano_traseiro_com.size} pixels")

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(16, 11))

ax1.imshow(plano_frontal_com)
ax1.set_title(f"Plano Frontal (Com Marcas para Gráfica) — {plano_frontal_com.width} x {plano_frontal_com.height} px", fontsize=13, fontweight='bold')
ax1.axis('on')
ax1.tick_params(labelsize=8)

ax2.imshow(plano_traseiro_com)
ax2.set_title(f"Plano Traseiro (Com Marcas para Gráfica) — {plano_traseiro_com.width} x {plano_traseiro_com.height} px", fontsize=13, fontweight='bold')
ax2.axis('on')
ax2.tick_params(labelsize=8)

plt.suptitle("Tritetraflexágono — Planificação Profissional para Gráfica (Com Marcas de Corte)", fontsize=15, fontweight='bold')
plt.tight_layout()
plt.show()
```

## 4. Comparação Lado a Lado (Sem Marcas vs Com Marcas)
Visão comparativa simultânea dos quatro planos.

```python
fig, axes = plt.subplots(2, 2, figsize=(18, 10))

axes[0, 0].imshow(plano_frontal_sem)
axes[0, 0].set_title(f"Plano Frontal — Sem Marcas ({plano_frontal_sem.width}x{plano_frontal_sem.height} px)", fontsize=11, fontweight='bold')
axes[0, 0].axis('off')

axes[0, 1].imshow(plano_frontal_com)
axes[0, 1].set_title(f"Plano Frontal — Com Marcas Gráfica ({plano_frontal_com.width}x{plano_frontal_com.height} px)", fontsize=11, fontweight='bold')
axes[0, 1].axis('off')

axes[1, 0].imshow(plano_traseiro_sem)
axes[1, 0].set_title(f"Plano Traseiro — Sem Marcas ({plano_traseiro_sem.width}x{plano_traseiro_sem.height} px)", fontsize=11, fontweight='bold')
axes[1, 0].axis('off')

axes[1, 1].imshow(plano_traseiro_com)
axes[1, 1].set_title(f"Plano Traseiro — Com Marcas Gráfica ({plano_traseiro_com.width}x{plano_traseiro_com.height} px)", fontsize=11, fontweight='bold')
axes[1, 1].axis('off')

plt.suptitle("Tritetraflexágono: Comparativo Geral das Planificações", fontsize=15, fontweight='bold')
plt.tight_layout()
plt.show()
```
