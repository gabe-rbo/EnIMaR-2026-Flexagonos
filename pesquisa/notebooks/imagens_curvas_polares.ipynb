## Gráficos de 6 curvas em coordenadas polares

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
#Catálogo de curvas
fig, axs = plt.subplots(2, 3, figsize=(12, 6), subplot_kw={'projection': 'polar'})

#Estrela: https://www.youtube.com/watch?v=xY9Be9HHRsk
ax = axs[0,0]

theta = np.linspace(0, 24*np.pi, 2000)
#r = np.sin(6*theta/5)**2 + np.sin(6*theta)**3
r = (np.exp(np.sin(theta)) - 2 * np.cos(4*theta) + np.sin((2*theta-np.pi)/24)**5)/2.3
#ax.plot(theta+(r<0)*np.pi, np.abs(r), lw = 1)
ax.plot(theta, r, lw = 1, color = 'tab:blue')
ax.plot(theta+np.pi, -r, lw = 1, color = 'tab:blue')

ax.set_rorigin(-0.5)  # <--- CENTRALIZA A BORBOLETA
ax.set_theta_offset(0)
#ax.set_rmax(2.2)
ax.axis("off")
ax.set_xticklabels([])
ax.set_yticklabels([])
ax.grid(False)
ax.set_title('"Borboleta" ($r = e^{sen(\\theta)} - 2cos(4\\theta) + sen^5((2\\theta-\pi)/24)}$)', va='bottom')

#Curva estrela
ax = axs[0,1]

theta = np.linspace(0*np.pi, 10*np.pi, 1000)
r = np.sin(1.2 * theta)**2 + np.cos(6*theta)**3
ax.plot(theta+(r<0)*np.pi, np.abs(r), lw = 1)

ax.set_rorigin(0)
#ax.set_rmax(2.1)
ax.axis("off")
ax.set_xticklabels([])
ax.set_yticklabels([])
ax.grid(False)
ax.set_title("Estrela ($r = sen^2(1.2\\theta) + cos^3(6\\theta)$)", va='bottom')

#Rosácea de três pétalas
#https://mathcurve.com/courbes2d.gb/rosace/rosace.shtml
ax = axs[0,2]

#Coordenadas dos pontos da rosácea
theta = np.linspace(0, 2 * np.pi,500)
r =  2 *  np.cos(3*theta)
ax.plot(theta, r, color = 'tab:blue')

ax.set_rorigin(0)
ax.set_rmax(2.1)
ax.axis("off")
ax.set_xticklabels([])
ax.set_yticklabels([])
ax.grid(False)
ax.set_title("Rosácea de três pétalas ($r = 2cos(3\\theta)$)", va='bottom')

#https://mathcurve.com/courbes2d.gb/rosace/rosace.shtml
#Rosáceas de quatro pétalas
ax = axs[1,0]

#Coordenadas dos pontos da rosácea
theta = np.linspace(0, 2 * np.pi,500)
r =  2 *  np.cos(2*theta)
ax.plot(theta+(r<0)*np.pi, np.abs(r), color = 'tab:blue')

r =  np.cos(2*theta)
ax.plot(theta+(r<0)*np.pi, np.abs(r), color = 'tab:blue')

ax.set_rorigin(0)
ax.set_rmax(2.1)
ax.axis("off")
ax.set_xticklabels([])
ax.set_yticklabels([])
ax.grid(False)
ax.set_title("Rosáceas ($r = 2cos(2\\theta), r = cos(2\\theta)$)", va='bottom')

#Flor de loto
ax = axs[1,1]

theta = np.linspace(0, 4*np.pi, 1000)
r = np.sin(theta) + np.sin(5*theta/2)**3

# Converter para coordenadas cartesianas
x = r * np.cos(theta)
y = r * np.sin(theta)

# DESLOCAR VERTICALMENTE (eixo Y)
deslocamento_y = -0.8  # Negativo desce, positivo sobe
y_deslocado = y + deslocamento_y

# Converter de volta para polares
r_novo = np.sqrt(x**2 + y_deslocado**2)
theta_novo = np.arctan2(y_deslocado, x)

# Garantir ângulos positivos
theta_novo = theta_novo % (2*np.pi)

# Plotar
ax.plot(theta_novo, r_novo, lw=1)


ax.set_rorigin(0)  
ax.axis("off")
ax.set_xticklabels([])
ax.set_yticklabels([])
ax.grid(False)
ax.set_title('"Flor de loto" ($r =  sen(\\theta) + sen(5\\theta/2)^3$)', va='bottom')

#Rosácea com fração no coeficiente
ax = axs[1,2]

theta = np.linspace(- np.pi, 20*np.pi/2.05 + np.pi, 1000)
r = 2* np.cos(2.05*theta)
ax.plot(theta+(r<0)*np.pi, np.abs(r))

ax.set_rorigin(0)
ax.set_rmax(2.1)
ax.axis("off")
ax.set_xticklabels([])
ax.set_yticklabels([])
ax.grid(False)
ax.set_title("Parte da Rosácea ($r = 2cos(2.05\\theta)$)", va='bottom')



plt.show()
```

```python
#Flor de loto
# Cria a figura e os eixos polares
fig, ax = plt.subplots(figsize=(6, 6), subplot_kw={'projection': 'polar'})

# Colore o fundo da imagem inteira (área FORA do círculo)
fig.set_facecolor('dodgerblue') 

# 2. Colore o fundo do gráfico polar (área DENTRO do círculo)
#ax.set_facecolor('ivory')

theta = np.linspace(0, 4*np.pi, 1000)
r = np.sin(theta) + np.sin(5*theta/2)**3

# Converter para coordenadas cartesianas
x = r * np.cos(theta)
y = r * np.sin(theta)

# DESLOCAR VERTICALMENTE (eixo Y)
deslocamento_y = -0.7  # Negativo desce, positivo sobe
y_deslocado = y + deslocamento_y

# Converter de volta para polares
r_novo = np.sqrt(x**2 + y_deslocado**2)
theta_novo = np.arctan2(y_deslocado, x)

# Garantir ângulos positivos
theta_novo = theta_novo % (2*np.pi)

# Plotar
ax.plot(theta_novo, r_novo, lw=2, color = 'lavender')
ax.fill(theta_novo, r_novo, lw=1, color = 'violet')

ax.set_rorigin(0)  
ax.axis("off")
ax.set_xticklabels([])
ax.set_yticklabels([])
ax.grid(False)

# Mostra o gráfico
plt.show()
```
